```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

show_img = cv2.imread('image.png')

img_rgb = cv2.cvtColor(show_img, cv2.COLOR_BGR2RGB)

plt.xticks([])
plt.yticks([])
plt.imshow(img_rgb)
```

```python
# ORIGINAL IMAGE TO BLUR IMAGE
k_size = 31

#Blur Image
mean_filter = np.ones((k_size,k_size), np.float32) / (k_size * k_size)
mean_filter2 = cv2.filter2D(img_rgb, -1, mean_filter)

#Dark Image
dark_img = cv2.convertScaleAbs(img_rgb, alpha=1.0, beta=-50)

#bright Image
bright_img = cv2.convertScaleAbs(img_rgb, alpha=1.0, beta=50)

#low contrast Image
low_contrast_img = cv2.convertScaleAbs(img_rgb, alpha=0.4, beta=0)

#High Contrast Image
high_contrast_img = cv2.convertScaleAbs(img_rgb, alpha=1.8, beta=0)

#Low Resolution
low_resolution_img = cv2.resize(img_rgb, (100, 50))

#High Resolution
high_resolution_img = cv2.resize(low_resolution_img, (1920, 1080))

#Noise
row,col,ch = img_rgb.shape
mean = 0
sigma = 50
gaus = np.random.normal(mean, sigma, (row, col, ch)).astype('uint8')

noise_img = cv2.add(img_rgb, gaus)


plt.subplot(3,3,1)
plt.imshow(img_rgb)
plt.title("Blur Image")
plt.axis("off")

plt.subplot(3,3,2)
plt.imshow(mean_filter2)
plt.title("Deblur Image")
plt.axis("off")

plt.subplot(3,3,3)
plt.imshow(dark_img)
plt.title("Dark Image")
plt.axis("off")

plt.subplot(3,3,4)
plt.imshow(bright_img)
plt.title("Bright Image")
plt.axis("off")

plt.subplot(3,3,5)
plt.imshow(low_contrast_img)
plt.title("Low Contrast")
plt.axis("off")

plt.subplot(3,3,6)
plt.imshow(high_contrast_img)
plt.title("High Contrast")
plt.axis("off")

plt.subplot(3,3,7)
plt.imshow(low_resolution_img)
plt.title("Low Resolution")
plt.axis("off")

plt.subplot(3,3,8)
plt.imshow(high_resolution_img)
plt.title("High Resolution")
plt.axis("off")

plt.subplot(3,3,9)
plt.imshow(noise_img)
plt.title("Noise")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# BLUR IMAGE TO ORIGINAL IMAGE
blur_img2 = cv2.imread('blur_img.png')

img_rgb = cv2.cvtColor(blur_img2, cv2.COLOR_BGR2RGB)

kernel = np.array([[0, -3, 0], [-3, 13, -3], [0, -3, 0]])
deblur = cv2.filter2D(img_rgb, -1, kernel)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Blur Image Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(deblur)
plt.title("Deblur Image")
plt.axis("off")
```
